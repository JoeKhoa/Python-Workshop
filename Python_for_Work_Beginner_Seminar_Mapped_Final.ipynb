## Slide 26 — Final mini-project 1: create a CSV report

This cell uses variables, data structures, if/else, loop, function, pandas, matplotlib and `to_csv()`.


```python
import pandas as pd
import matplotlib.pyplot as plt

students = [
    {"name": "Maya", "score": 85},
    {"name": "Leo", "score": 62},
    {"name": "Ava", "score": 45},
]

def get_status(score):
    if score >= 50:
        return "Pass"
    else:
        return "Review"

for student in students:
    student["status"] = get_status(student["score"])

df = pd.DataFrame(students)
df.to_csv("student_scores.csv", index=False)

print(df)
print("Average:", df["score"].mean())

df.plot(x="name", y="score", kind="bar")
plt.title("Student Score Report")
plt.show()
```

## Slide 27 — Final mini-project 2: read, process and visualise data

Read the CSV from project 1, calculate answers, filter rows and plot the result.


```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("student_scores.csv")

print(df.head())
print(df["status"].value_counts())

average_score = df["score"].mean()
highest_score = df["score"].max()

print("Average:", average_score)
print("Highest:", highest_score)

passed_students = df[df["status"] == "Pass"]
print(passed_students)

df.plot(x="name", y="score", kind="bar")
plt.axhline(y=50, linestyle="--")
plt.title("Pass Line and Student Scores")
plt.xlabel("Student")
plt.ylabel("Score")
plt.show()
```
